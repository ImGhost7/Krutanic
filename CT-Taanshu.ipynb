## Section 1: Basic Python

### 1. FizzBuzz

```python
for i in range(1, 51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

### 2. Leap Year Logic

```python
def is_leap_year(year):
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
```

### 3. Identity vs Equality

```python
L1 = [1, 2, 3]
L2 = [1, 2, 3]

print(L1 == L2)
print(L1 is L2)
```

### 4. Bitwise Swap

```python
a = 10
b = 20

a = a ^ b
b = a ^ b
a = a ^ b

print(a, b)
```

## Section 2: Control Flow & Loops

### 5. Prime Number Finder

```python
n = 2
count = 0

while count < 10:
    prime = True
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            prime = False
            break
    if prime:
        print(n)
        count += 1
    n += 1
```

### 6. The Loop Breaker

```python
for i in range(1, 21):
    if i % 4 == 0:
        pass
    if i == 13:
        continue
    if i == 18:
        break
    print(i)
```

### 7. Nested Conditionals

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    return "Fail"

print(grade(85))
```

### 8. Reverse an Integer

```python
n = 12345
rev = 0

while n > 0:
    rev = rev * 10 + n % 10
    n //= 10

print(rev)
```

## Section 3: Data Structures

### 9. Runner-Up Score

```python
scores = [10, 20, 20, 8, 15]
highest = runner_up = float("-inf")

for score in scores:
    if score > highest:
        runner_up = highest
        highest = score
    elif highest > score > runner_up:
        runner_up = score

print(runner_up)
```

### 11. List Intersection

```python
L1 = [1, 2, 3, 4]
L2 = [3, 4, 5, 6]

common = [x for x in L1 if x in L2]
common
```

## Section 4: Dictionaries & Comprehensions

### 13. Character Frequency Counter

```python
text = input("Enter a string: ")
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

frequency
```

### 14. List Comprehension

```python
squares = [x ** 2 for x in range(1, 21) if x % 2 == 0]
squares
```

### 15. Dictionary Comprehension

```python
data = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
result = {k: v for k, v in data.items() if v > 2}
result
```
