# Classification Algorithms Assignment


## 1. Linear Classification Algorithms

Algorithms:
1. Logistic Regression
2. Perceptron
3. Ridge Classifier
4. Linear Discriminant Analysis (LDA)
5. Quadratic Discriminant Analysis (QDA)
6. Linear Support Vector Machine (Linear SVM)
7. Stochastic Gradient Descent (SGD Classifier)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time

from sklearn.datasets import load_iris, load_breast_cancer, load_digits, load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

RANDOM_STATE = 42
sns.set_theme(style="whitegrid")

def evaluate_model(model, X_train, X_test, y_train, y_test, name):
    start = time.time()
    model.fit(X_train, y_train)
    train_time = time.time() - start
    pred = model.predict(X_test)
    print(f"\n{name}")
    print("=" * len(name))
    print("Accuracy :", round(accuracy_score(y_test, pred), 4))
    print("Precision:", round(precision_score(y_test, pred, average="weighted", zero_division=0), 4))
    print("Recall   :", round(recall_score(y_test, pred, average="weighted", zero_division=0), 4))
    print("F1-score :", round(f1_score(y_test, pred, average="weighted", zero_division=0), 4))
    print("Training time:", round(train_time, 4), "seconds")
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, pred))
    return model, pred, train_time

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
```

```python
from sklearn.linear_model import LogisticRegression, Perceptron, RidgeClassifier, SGDClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.svm import LinearSVC

linear_models = [
    ("Logistic Regression", make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))),
    ("Perceptron", make_pipeline(StandardScaler(), Perceptron(random_state=RANDOM_STATE))),
    ("Ridge Classifier", make_pipeline(StandardScaler(), RidgeClassifier())),
    ("LDA", LinearDiscriminantAnalysis()),
    ("QDA", make_pipeline(StandardScaler(), QuadraticDiscriminantAnalysis())),
    ("Linear SVM", make_pipeline(StandardScaler(), LinearSVC(random_state=RANDOM_STATE, max_iter=5000))),
    ("SGD Classifier", make_pipeline(StandardScaler(), SGDClassifier(random_state=RANDOM_STATE, max_iter=2000, tol=1e-3)))
]

results = []
trained_models = {}

for name, model in linear_models:
    fitted, pred, train_time = evaluate_model(model, X_train, X_test, y_train, y_test, name)
    trained_models[name] = fitted
    results.append({
        "Algorithm": name,
        "Category": "Linear Classification",
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, average="weighted", zero_division=0),
        "Recall": recall_score(y_test, pred, average="weighted", zero_division=0),
        "F1-score": f1_score(y_test, pred, average="weighted", zero_division=0),
        "Training Time (s)": train_time
    })

pd.DataFrame(results)
```

```python
from sklearn.neural_network import MLPClassifier

mlp = make_pipeline(
    StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=500, random_state=RANDOM_STATE, early_stopping=True)
)
fitted, pred, train_time = evaluate_model(mlp, X_train, X_test, y_train, y_test, "Multilayer Perceptron (MLP)")
results.append({
    "Algorithm":"MLP","Category":"Neural Network",
    "Accuracy":accuracy_score(y_test,pred),
    "Precision":precision_score(y_test,pred,average="weighted",zero_division=0),
    "Recall":recall_score(y_test,pred,average="weighted",zero_division=0),
    "F1-score":f1_score(y_test,pred,average="weighted",zero_division=0),
    "Training Time (s)":train_time
})
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("PyTorch device:", device)
```

```python
digits = load_digits()
X_img = torch.tensor(digits.images[:, None, :, :], dtype=torch.float32) / 16.0
y_img = torch.tensor(digits.target, dtype=torch.long)

img_train, img_test, y_img_train, y_img_test = train_test_split(
    X_img, y_img, test_size=0.2, random_state=RANDOM_STATE, stratify=y_img
)

class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(32 * 2 * 2, 64), nn.ReLU(),
            nn.Linear(64, 10)
        )
    def forward(self, x):
        return self.net(x)

cnn = SimpleCNN().to(device)
loader = DataLoader(TensorDataset(img_train, y_img_train), batch_size=64, shuffle=True)
optimizer = torch.optim.Adam(cnn.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(3):
    cnn.train()
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = loss_fn(cnn(xb), yb)
        loss.backward()
        optimizer.step()

cnn.eval()
with torch.no_grad():
    pred = cnn(img_test.to(device)).argmax(1).cpu()
print("CNN Accuracy:", round(accuracy_score(y_img_test, pred), 4))
```

```python
torch.manual_seed(RANDOM_STATE)

n_samples, seq_len, input_size = 1000, 20, 4
X_seq = torch.randn(n_samples, seq_len, input_size)
y_seq = (X_seq.mean(dim=(1, 2)) > 0).long()

seq_train, seq_test, y_seq_train, y_seq_test = train_test_split(
    X_seq, y_seq, test_size=0.2, random_state=RANDOM_STATE, stratify=y_seq
)

class SequenceClassifier(nn.Module):
    def __init__(self, cell="RNN"):
        super().__init__()
        if cell == "RNN":
            self.rnn = nn.RNN(input_size, 32, batch_first=True)
        elif cell == "LSTM":
            self.rnn = nn.LSTM(input_size, 32, batch_first=True)
        else:
            self.rnn = nn.GRU(input_size, 32, batch_first=True)
        self.fc = nn.Linear(32, 2)

    def forward(self, x):
        out, _ = self.rnn(x)
        return self.fc(out[:, -1, :])

for cell in ["RNN", "LSTM", "GRU"]:
    model = SequenceClassifier(cell).to(device)
    loader = DataLoader(TensorDataset(seq_train, y_seq_train), batch_size=64, shuffle=True)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
    loss_fn = nn.CrossEntropyLoss()

    for epoch in range(3):
        model.train()
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            loss = loss_fn(model(xb), yb)
            loss.backward()
            optimizer.step()

    model.eval()
    with torch.no_grad():
        pred = model(seq_test.to(device)).argmax(1).cpu()
    print(f"{cell} Accuracy:", round(accuracy_score(y_seq_test, pred), 4))
```

```python
class SimpleTransformer(nn.Module):
    def __init__(self):
        super().__init__()
        d_model = 32
        self.proj = nn.Linear(input_size, d_model)
        layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=4, batch_first=True)
        self.encoder = nn.TransformerEncoder(layer, num_layers=2)
        self.fc = nn.Linear(d_model, 2)

    def forward(self, x):
        x = self.proj(x)
        x = self.encoder(x)
        return self.fc(x.mean(dim=1))

transformer = SimpleTransformer().to(device)
loader = DataLoader(TensorDataset(seq_train, y_seq_train), batch_size=64, shuffle=True)
optimizer = torch.optim.Adam(transformer.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(3):
    transformer.train()
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = loss_fn(transformer(xb), yb)
        loss.backward()
        optimizer.step()

transformer.eval()
with torch.no_grad():
    pred = transformer(seq_test.to(device)).argmax(1).cpu()
print("Transformer Accuracy:", round(accuracy_score(y_seq_test, pred), 4))
```
